```python
df['Date'].fillna(df['Date'].mode(), inplace=True)

df['Open'].fillna(df['Open'].mean(), inplace=True)

df['High'].fillna(df['High'].mean(), inplace=True)

df['Low'].fillna(df['Low'].mean(), inplace=True)

df['Close'].fillna(df['Close'].mean(), inplace=True)

df['Predicted_Close'].fillna(df['Predicted_Close'].mean(), inplace=True)

df['Signal'].fillna(df['Signal'].mode(), inplace=True)

df['Confidence'].fillna(df['Confidence'].mean(), inplace=True)
```

```python
df['Volume'] = pd.to_numeric(df['Volume'], errors='coerce')

df['Volume'].fillna(df['Volume'].mean(), inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df['Date'].fillna(df['Date'].median(), inplace=True)
```

```python
df['Signal'].fillna(df['Signal'].mode()[0], inplace=True)
```
